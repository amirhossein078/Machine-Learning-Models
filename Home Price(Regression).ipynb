```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
%matplotlib inline
```

```python
df = pd.read_csv("1632300362534233.csv")
df.dtypes
df.head()
```

```python
df = df.dropna(subset=['Address'])
df['Area'] = df['Area'].astype(str).str.replace(',','').str.strip()
df['Area'] = pd.to_numeric(df['Area'], errors = 'coerce')
df = df[(df['Area'] >= 30 ) & (df['Area'] <= 1000)]
df = df.dropna(subset=['Price','Room'])
df = df[(df['Price'] > 0)]
df = pd.get_dummies(df,columns=["Address"], dtype=int)
df.shape
```

```python
x = df.drop(columns=["Price","Price(USD)"])
y = df["Price"]
```

```python
from sklearn.preprocessing import LabelEncoder
le = LabelEncoder()
for i in ['Parking','Warehouse','Elevator']:
    df[i] = le.fit(['False','True']).transform(df[i])
```

```python
df.head()
```

```python
from sklearn.model_selection import train_test_split
x_train, x_test, y_train, y_test = train_test_split(x,y, test_size=0.2,random_state=6)
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score, mean_absolute_error
reg = LinearRegression()
reg.fit(x_train, y_train)
y_pre = reg.predict(x_test)
```

```python
print(r2_score(y_test, y_pre))
mean_absolute_error(y_test,y_pre)
```
