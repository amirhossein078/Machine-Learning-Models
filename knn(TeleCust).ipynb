```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn import preprocessing
from sklearn.model_selection import train_test_split
%matplotlib inline
```

```python
df = pd.read_csv("csv/teleCust1000t.csv")
df.head()
```

```python
x = df[['region','tenure','marital','address','income','ed','employ','retire','gender','reside',]].values
y = df['custcat'].values
```

```python
scaler = preprocessing.StandardScaler().fit(x)
x = scaler.transform(x.astype(float))
```

```python
x_train, x_test, y_train, y_test = train_test_split(x,y, test_size=0.2, random_state=4)
print(x_train.shape, y_train.shape)
print(x_test.shape, y_test.shape)
```

```python
from sklearn.neighbors import KNeighborsClassifier
```

```python
k = 4
neigh = KNeighborsClassifier(n_neighbors = k).fit(x_train, y_train)
```

```python
y_ = neigh.predict(x_test)
print(y[0:5])
print(y_[0:5])
```

```python
from sklearn.metrics import accuracy_score
print(accuracy_score(y_train, neigh.predict(x_train)))
print(accuracy_score(y_test, neigh.predict(x_test)))
```

```python
ks = 10
mean_acc = np.zeros((ks - 1))
std_acc = np.zeros((ks - 1))
for n in range(1,ks):
    neigh = KNeighborsClassifier(n_neighbors = n).fit(x_train, y_train)
    yhat = neigh.predict(x_test)

    mean_acc[n-1] = accuracy_score(y_test, yhat)
    std_acc[n-1]=np.std(yhat==y_test)/np.sqrt(yhat.shape[0])
mean_acc
```

```python
print("accuracy = ",mean_acc.max(),"and k =",mean_acc.argmax()+1)
```
