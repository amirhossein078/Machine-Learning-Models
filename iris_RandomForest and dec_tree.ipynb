```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
```

```python
df = pd.DataFrame(load_iris().data, columns=load_iris().feature_names)
df['target'] = load_iris().target
df
```

```python
x = df.drop(columns=['target'])
y = df[["target"]]
```

```python
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2,random_state=42)
x_train.shape, x_test.shape, y_train.shape, y_test.shape
```

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.tree import DecisionTreeClassifier
```

```python
model = RandomForestClassifier(criterion='entropy', max_depth=4, random_state=42)
model.fit(x_train, y_train)
predict = model.predict(x_test)
```

```python
from sklearn import metrics
print("DecisionTrees's Accuracy: ", metrics.accuracy_score(y_test, predict))
```

```python
cm = metrics.confusion_matrix(y_test, predict)
display = metrics.ConfusionMatrixDisplay(confusion_matrix=cm, display_labels=load_iris().target_names)
display.plot()
plt.show()
```

```python
model1 = DecisionTreeClassifier(criterion='entropy', max_depth=4)
model1.fit(x_train, y_train)
```

```python
predict1 = model1.predict(x_test)
```
